```python
def FriTexAna(text, keywords): # Pending, change the keyword implementation so that they are chosen automatically from the text.
  if type(text)==str:
    text=Text(nltk.word_tokenize(text))
  # textp=Text(nltk.word_tokenize(text))
  a=len(text)
  c=len(set(text))
  #text=Text(nltk.word_tokenize(text))
  print("\n Word Count", a)
  print("\n Unique Word Count", c)
  print("\n Word Diversity Index ", a/c)
  #print("\n Word Diversity Index 2", b/c)



  print("\n\nConcordance of", str(keywords[0]), ":\n")
  c0=text.concordance(keywords[0])
  print("\n\nConcordance of", str(keywords[1]), ":\n")
  c1=text.concordance(keywords[1])
  print("\n\nConcordance of", str(keywords[2]), ":\n")
  c2=text.concordance(keywords[2])

  print("\n\nSimilar words to", str(keywords[0]), ":\n")
  s0=text.similar(keywords[0])
  print("\n\nSimilar words to", str(keywords[1]), ":\n")
  s1=text.similar(keywords[1])
  print("\n\nSimilar words to", str(keywords[2]), ":\n")
  s2=text.similar(keywords[2])
  print("\n\nText Collocations")
  c=text.collocations()
  print("\n\nCommon contexts of keywords", str(keywords[0]), "and", str(keywords[1]), "\n")
  cc01=text.common_contexts([keywords[0], keywords[0]])
  print("\n\nCommon contexts of keywords", str(keywords[0]), "and", str(keywords[2]), "\n")
  cc02=text.common_contexts([keywords[1], keywords[1]])
  print("\n\nCommon contexts of keywords", str(keywords[1]), "and", str(keywords[2]), "\n")
  cc12=text.common_contexts([keywords[2], keywords[2]])
  # How did we make sure plots didn't blend together again?
  print("\n\nDispersion plot:")
  fig = plt.figure()
  d=text.dispersion_plot(keywords)
  plt.show()
  fig.savefig('DisPlot.png', bbox_inches = "tight")
  # How did we make sure plots didn't blend together again?
  print("\n\nWord Frequency Distribution plot:")
  fig = plt.figure()
  fdist = FreqDist(text)
  fdist.plot(title = "Word frequency distribution", cumulative = False)
  plt.show()
  fig.savefig('FreqDist.png', bbox_inches = "tight")

  # How did we make sure plots didn't blend together again?
  print("\n\nCummulative Word Frequency Distribution")
  fig = plt.figure()
  fdist.plot(title = "Cumulative word frequency distribution", cumulative = True)
  plt.show()
  fig.savefig('AccFreqDist.png', bbox_inches = "tight")


  # Estimate the beta and k parameters for the Heap's Law
  print("\n\nHeap Law:")
  x=[]
  y=[]
  for i in range(1, len(text),100):
    x.append(i)
    y.append(len(set(text[0:i])))


  popt, pcov =sp.optimize.curve_fit(function, x, y, p0=[1,1])
  print("The fitting k and beta values", popt)
  fit_fn = []
  for i in x:
    fit_fn.append(function(i, popt[0], popt[1]))

  fig=plt.figure()
  plt.scatter(x, y)          # blue
  plt.scatter(x, fit_fn)  # orange
  plt.grid()
  plt.show()
  # Find Hapaxes
  fdist = FreqDist(text)
  print("\n\nHapaxes:")
  fdist.hapaxes()

```
